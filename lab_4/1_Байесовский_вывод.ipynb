# Задача 1

Пусть $X_1, X_2, \ldots, X_n$ — выборка из экспоненциального распределения с параметром $\lambda$. Найти оценку максимального правдоподобия параметра $\lambda$, сравнить ее с байесовской оценкой (MAP и математическое ожидание апостреорного распределения), подобрав сопряженное распределение. Сравнить полученные байесовские оценки с оценкой MLE. Найти предсказательное распределение

### 1. Оценка максимального правдоподобия (MLE)

У нас есть выборка $X = (X_1, \ldots, X_n)$ из экспоненциального распределения.
Плотность вероятности для одного наблюдения:
$$f(x|\lambda) = \lambda e^{-\lambda x}$$

Запишем функцию правдоподобия.
Это вероятность получить наши данные. Так как наблюдения независимы, просто перемножаем их плотности:
$$L(\lambda) = \prod_{i=1}^n \lambda e^{-\lambda x_i} = \lambda^n e^{-\lambda \sum x_i}$$

Обозначим сумму всех иксов как $S = \sum_{i=1}^n x_i$. Тогда:
$$L(\lambda) = \lambda^n e^{-\lambda S}$$

Искать максимум произведения сложно, проще искать максимум логарифма:
$$l(\lambda) = \ln L(\lambda) = n \ln \lambda - \lambda S$$

Берем производную по $\lambda$ и приравниваем к нулю:
$$\frac{d l}{d \lambda} = \frac{n}{\lambda} - S = 0$$

Теперь находим $\lambda$:
$$\hat{\lambda}_{MLE} = \frac{n}{S} = \frac{n}{\sum_{i=1}^n x_i} = \frac{1}{\bar{X}}$$

### 2. Подбор сопряженного априорного распределения

В байесовском подходе мы считаем $\lambda$ случайной величиной. Нам нужно выбрать для нее априорное распределение $p(\lambda)$.

Посмотрим на формулу правдоподобия, которую мы получили выше:
$$L(\lambda) \propto \lambda^n e^{-\lambda S}$$

Такую же структуру имеет Гамма-распределение. Значит, оно является сопряженным.

Пусть априорное распределение $\lambda \sim \text{Gamma}(\alpha, \beta)$. Его плотность:
$$p(\lambda) = \frac{\beta^\alpha}{\Gamma(\alpha)} \lambda^{\alpha-1} e^{-\beta \lambda}$$

где $\alpha$ и $\beta$ — параметры, которые мы задаем до опыта (наши начальные знания).


### 3. Апостериорное распределение

Мы получили новые данные $X$. По формуле Байеса обновляем наше мнение о $\lambda$:
$$p(\lambda | X) = p(X | \lambda) \cdot p(\lambda)$$
$$\text{Апостериорное} \propto \text{Правдоподобие} \times \text{Априорное}$$

Перемножаем формулы:
$$p(\lambda | X) = (\lambda^n e^{-\lambda S}) \cdot (\lambda^{\alpha-1} e^{-\beta \lambda})$$

Складываем степени при $\lambda$ и показатели экспоненты:
$$p(\lambda | X) = \lambda^{(n + \alpha) - 1} e^{-\lambda (S + \beta)}$$

Мы видим, что получилась та же формула Гамма-распределения, но с новыми параметрами.

Значит, апостериорное распределение — это тоже Гамма, но с параметрами:
*   $\alpha_{new} = \alpha + n$
*   $\beta_{new} = \beta + S = \beta + \sum x_i$

$$p(\lambda | X) = \text{Gamma}(\alpha + n, \beta + \sum x_i)$$


### 4. Байесовские оценки и сравнение с MLE

Теперь вытащим из полученного распределения конкретные числа.

**1. Оценка MAP**
Это мода апостериорного распределения. Для Гамма-распределения мода равна $\frac{\alpha_{new}-1}{\beta_{new}}$:
$$\hat{\lambda}_{MAP} = \frac{\alpha + n - 1}{\beta + \sum x_i}$$

**2. Оценка мат. ожидания**
Это среднее значение апостериорного распределения:
$$\hat{\lambda}_{Mean} = \frac{\alpha + n}{\beta + \sum x_i}$$


Байесовские оценки похожи на MLE, но к числителю и знаменателю добавляются наши априорные параметры $\alpha$ и $\beta$.
*   Если данных ($n$) очень много, то $\alpha$ и $\beta$ почти не влияют $\to$ оценки совпадают.
*   Если данных мало, байесовская оценка "тяготеет" к нашим априорным знаниям.


### 5. Предсказательное распределение

Мы хотим предсказать вероятность появления нового значения $x_{new}$, учитывая уже имеющиеся данные. Для этого мы усредняем предсказание по всем возможным $\lambda$:

$$p(x_{new} | X) = \int_0^\infty p(x_{new} | \lambda) p(\lambda | X) d\lambda$$

Подставляем экспоненциальное (для икса) и полученное Гамма (для лямбды):
$$p(x_{new} | X) = \int_0^\infty (\lambda e^{-\lambda x_{new}}) \cdot \left( \frac{\beta_{new}^{\alpha_{new}}}{\Gamma(\alpha_{new})} \lambda^{\alpha_{new}-1} e^{-\beta_{new} \lambda} \right) d\lambda=$$
$$ = \frac{\beta_{new}^{\alpha_{new}}}{\Gamma(\alpha_{new})} \int_0^\infty \lambda^{(\alpha_{new}+1)-1} e^{-\lambda(\beta_{new} + x_{new})} d\lambda$$

Интеграл внутри — это в точности кусок от другого Гамма-распределения, и он равен $\frac{\Gamma(\alpha_{new}+1)}{(\beta_{new} + x_{new})^{\alpha_{new}+1}}$.

Используем свойство $\Gamma(z+1) = z\Gamma(z)$ и сокращаем:
$$p(x_{new} | X) = \frac{\alpha_{new} \cdot (\beta_{new})^{\alpha_{new}}}{(\beta_{new} + x_{new})^{\alpha_{new} + 1}}$$

Это формула распределения Ломакса (или Парето II рода).


# Задача 2

**Мультиномиальное распределение**

Пусть проводится серия из $n$ испытаний и в результате каждого испытания происходит ровно одно событие из набора $A_1, A_2, \dots, A_m$, причем вероятности этих событий равны соответственно $\mathsf{p}_1, \mathsf{p}_2, \dots, \mathsf{p}_m$, причем
$$\sum_{i=1}^{m}\mathsf{p}_i = 1.$$

Тогда совместное распределение величин $X_1, X_2, \dots, X_m$, где $X_k$ — число наступлений события $A_k$ в серии из $n$ испытаний, задается вероятностями

$$
\mathsf{P}\left(X_1 = n_1, \dots, X_m = n_m, \right) = \frac{n!}{n_1!\dots n_m!}\mathsf{p}_1^{n_1}\dots \mathsf{p}_m^{n_m},
$$

где $n_1, n_2, \dots, n_m$ — произвольный набор целых неотрицательных чисел, таких что

$$\sum_{i=1}^m n_i = n.$$

Произведите байесовский вывод для мультиномиального распределения: найдите апостериорное распределение, используя в качестве сопоряженного распределения к правдоподобию [распределение Дирихле](https://ru.wikipedia.org/wiki/%D0%A0%D0%B0%D1%81%D0%BF%D1%80%D0%B5%D0%B4%D0%B5%D0%BB%D0%B5%D0%BD%D0%B8%D0%B5_%D0%94%D0%B8%D1%80%D0%B8%D1%85%D0%BB%D0%B5), найдите предсказательное распределение. Объясните результат.

### Апостериорное распределение

По формуле Байеса умножаем априорное на правдоподобие:

$$P(\mathsf{p} | X) = L(\mathsf{p}) \cdot Dir(\mathsf{p} | \boldsymbol{\alpha})$$

$$P(\mathsf{p} | X) = \left( \prod_{k=1}^m \mathsf{p}_k^{x_k} \right) \cdot \left( \prod_{k=1}^m \mathsf{p}_k^{\alpha_k - 1} \right)$$

Снова просто складываем показатели степеней для каждого $\mathsf{p}_k$:

$$P(\mathsf{p} | X) = \prod_{k=1}^m \mathsf{p}_k^{(x_k + \alpha_k) - 1}$$

Мы видим, что формула сохранила свой вид (произведение степеней). Значит, апостериорное распределение — это снова распределение Дирихле, но с обновленными параметрами.


$$P(\mathsf{p} | X) = Dir(\alpha_1 + x_1, \alpha_2 + x_2, \dots, \alpha_m + x_m)$$


### Предсказательное распределение

Матожидание для Дирихле $Dir(\boldsymbol{\alpha}')$ известно и равно:
$$E[\mathsf{p}_k] = \frac{\alpha'_k}{\sum_{j=1}^m \alpha'_j}$$

Подставляем наши обновленные параметры $\alpha'_k = \alpha_k + x_k$:

$$P(X_{new} = k | X) = \frac{\alpha_k + x_k}{\sum_{j=1}^m (\alpha_j + x_j)} = \frac{\alpha_k + x_k}{N + \sum \alpha_j}$$
где $N$ — общее число бросков.
